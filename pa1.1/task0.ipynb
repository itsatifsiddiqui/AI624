### Important Notes

#### 1. Memory Profiling Limitation
**Memory usage profiling only works accurately on CPU devices.**

- PyTorch's profiler can only track CPU memory usage accurately
- When using CUDA or MPS, memory is allocated on GPU (not tracked by CPU profiler)
- Results show 0.00 MB on GPU devices

**Solution**: Run profiling with `DEVICE = 'cpu'` for memory metrics
- CIFAR-10 CPU results: Peak = 270.03 MB, Average = 78.47 MB
- CIFAR-100 CPU results: Peak = 270.03 MB, Average = 78.48 MB

#### 2. Energy Consumption
**Energy measurement not available (requires Linux with RAPL support).**

- pyJoules requires direct hardware access via Intel RAPL
- RAPL is not available on:
  - macOS (Apple Silicon or Intel)
  - Windows
  - Virtual machines (including Google Colab)
- Official documentation: https://github.com/powerapi-ng/pyJoules/blob/master/docs/index.rst#limitation
